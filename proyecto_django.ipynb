```python
#  Clase de Campaña para Influencers
class Campaña:
    def __init__(self, nombre_empresa, presupuesto):
        self.nombre_empresa = nombre_empresa
        self.presupuesto = presupuesto
        self.estado = "Protegido (Escrow)"

    def mostrar_detalle(self):
        return f"Proyecto: {self.nombre_empresa} | Presupuesto: ${self.presupuesto} | Estado: {self.estado}"

proyecto_ebac = Campaña("Sebastian Corp", 500)
print(proyecto_ebac.mostrar_detalle())
```
